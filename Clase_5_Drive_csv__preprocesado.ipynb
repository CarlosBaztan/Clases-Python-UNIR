# Preprocesado

```python
import pandas as pd

#Simulamos un dataset "sucio"
datos_sucios ={
    "ID Cliente ": [1,2,2,4,5], #Fijaos el espacio al final de ID Cliente y fijaros tambien en que el 2 está repetido
    "NombreCompleto": ["Ana", "Luis", "Luis", "Pedro", None], #Volvemos a tener Luis duplicado y no tenemos un registro
    "Edad_Años": [25,30,30,None, 45], #No tenemos la edad de Pedro
    "Columna":["X1", "X2", "X2", "X3", "X4"] #Columna que no aporta información
}
df_limpieza = pd.DataFrame(datos_sucios)
df_limpieza
```

Estandarizar/Renonmbrar columnas

```python
#A través del rename() pasandole un diccionario de {"nombre_viejo": "nombre_nuevo"}

df_limpieza = df_limpieza.rename(columns = {
    "ID Cliente ":"id_cliente",
    "NombreCompleto":"nombre",
    "Edad_Años": "edad",
    "Columna": "columna"
})
df_limpieza
```

Eliminar lo que no aporta valor

```python
#A través de drop(). Se utiliza axis=1 para borrar una columna. Si quisieramos borrar una fila, usaríamos axis=0.
df_limpieza = df_limpieza.drop("columna", axis=1)
df_limpieza
```

Eliminar duplicados

```python
print(f"Filas antes de borrar duplicados: {len(df_limpieza)}")
df_limpieza=df_limpieza.drop_duplicates()
print(f"Filas despues de borrar duplicados: {len(df_limpieza)}")
df_limpieza
```

El siguiente paso es importante porque crea una copia independiente del DataFrame original en pandas, lo que permite trabajar y hacer modificaciones sobre df_limpieza_2 sin afectar a df_limpieza. Si no usaramos .copy(), ambas variables apuntarían al mismo objeto en memoria, por lo que cualquier cambio en una también modificaría la otra, pudiendo provocar errores o pérdida de datos originales durante el proceso de limpieza.

```python
df_limpieza_2=df_limpieza.copy()
df_limpieza_2
```

Imputar valores nulos

```python
#Imputar con un valor
df_limpieza_2["nombre"] = df_limpieza_2["nombre"].fillna("Desconocido")
df_limpieza_2
```

```python
#Nivel 1: Media
media_global = df_limpieza_2["edad"].mean()
df_limpieza_2["edad"] = df_limpieza_2["edad"].fillna(media_global)
df_limpieza_2
```

```python
#Nivel 2. Media por grupos, se utiliza la combinación groupby() + transform()

datos_sucios = {
    'ID Cliente ': [1, 2, 2, 4, 5],
    'NombreCompleto': ['Ana', 'Luis', 'Luis', 'Pedro', None],
    'Edad_Años': [25, 25, 80, None, 45],
     "Segmento": ["VIP", "NO VIP", "VIP", "VIP", "NO VIP"],
    'Columna_Inutil_Sistema': ['X1', 'X2', 'X2', 'X3', 'X4']
}
df_limpieza = pd.DataFrame(datos_sucios)

df_limpieza = df_limpieza.rename(columns={
    'ID Cliente ': 'id_cliente',
    'NombreCompleto': 'nombre',
    'Edad_Años': 'edad',
    'Segmento': 'segmento',
    'Columna_Inutil_Sistema': 'codigo_tecnico'
})
df_limpieza = df_limpieza.drop("codigo_tecnico", axis=1)
df_limpieza["nombre"] = df_limpieza["nombre"].fillna("Desconocido")
df_limpieza
```

```python
# Imputar el missing de edad usando la media de cada segmento
df_limpieza["edad"] = df_limpieza["edad"].fillna(
    df_limpieza.groupby("segmento")["edad"].transform("mean")
)
df_limpieza
```

```python
# Nivel 3. KnnImputer
from sklearn.impute import KNNImputer

datos_clientes = {
    'id_cliente': [1, 2, 3, 4, 5],
    'nombre': ['Ana', 'Luis', 'Carlos', 'Pedro', 'Marta'],
    'ingresos_anuales': [25000, 27000, 80000, 82000, 24000], # Fijaos: Ana, Luis y Marta ganan poco. Carlos y Pedro ganan mucho.
    'gasto_mensual': [150, 180, 800, None, 140]              # Pedro es un misterio (None)
}
df_ml = pd.DataFrame(datos_clientes)
df_ml
```

```python
#llamamos a la función de knn
algoritmo_knn=KNNImputer(n_neighbors=1)

#Agrupamos todas las variables numericas. porque KNN solo entiende de NUMEROS
columnas_numericas=["ingresos_anuales", "gasto_mensual"]

df_ml[["ingresos_anuales","gasto_mesual_ml"]] = algoritmo_knn.fit_transform(df_ml[columnas_numericas])
df_ml
```
